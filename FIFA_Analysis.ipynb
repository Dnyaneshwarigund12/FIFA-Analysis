```python
# fifa_analysis.py
"""
FIFA 20 Player Dataset Analysis

This script analyzes the FIFA 20 player dataset (`players_20.csv`) to explore player attributes,
including nationality, club, shooting, defending, and wages. The dataset contains 18,278 players
with 104 attributes. The analysis includes data exploration, visualizations, and club-specific
insights for Real Madrid.
"""

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np















```

```python
# Load the dataset
print("Loading dataset...")
fifa = pd.read_csv('/content/players_20 (1).csv')
```

```python
# Display first 5 rows
print("\nFirst 5 rows of the dataset:")
print(fifa.head())
```

```python

# Check dataset shape
print("\nDataset shape:", fifa.shape)
```

```python
# Display summary statistics for numerical columns
print("\nSummary statistics:")
print(fifa.describe())
```

```python
# List all columns in the dataset
print("\nDataset columns:")
for col in fifa.columns:
    print(col)
```

```python
# Data Cleaning
# Handle missing values in key columns and clean positional ratings
print("\nChecking for missing values in key columns:")
print(fifa[['shooting', 'defending']].isna().sum())
```

```python
# Impute missing values with median
fifa['shooting'].fillna(fifa['shooting'].median(), inplace=True)
fifa['defending'].fillna(fifa['defending'].median(), inplace=True)
```

```python
# Function to clean positional ratings (e.g., '80+3' to 80)
def clean_position_rating(rating):
    """Convert positional rating strings (e.g., '80+3') to integers."""
    if isinstance(rating, str):
        return int(rating.split('+')[0])
    return rating
```

```python
# Apply cleaning to positional columns
for col in ['ls', 'st', 'cb', 'rb']:
    fifa[col] = fifa[col].apply(clean_position_rating)
print("\nPositional ratings cleaned for columns: ls, st, cb, rb")

```

```python
# Nationality Analysis
# Get top 10 nationalities by player count
print("\nTop 10 nationalities by player count:")
print(fifa['nationality'].value_counts()[:10])

```

```python
# Get top 5 nationalities
top_nationalities = fifa['nationality'].value_counts()[:5]
print("\nTop 5 nationalities:")
print(top_nationalities)

```

```python
 #Player Performance Analysis
# Function to display top players by a given attribute
def top_players_by_attribute(df, attribute, n=5):
    """Display top n players sorted by the specified attribute."""
    return df[['short_name', attribute, 'nationality', 'club']].sort_values(by=attribute, ascending=False).head(n)
```
